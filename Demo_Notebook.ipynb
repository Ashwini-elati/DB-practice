```python
%sql
Use catalog my_demo_catalog
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS my_demo_catalog.demo_schema;
```

```python
data = [
    (101, "Ravi", "Hyderabad", 30),
    (102, "Sita", "Chennai", 25),
    (103, "John", "Bangalore", 35),
    (104, "Priya", "Mumbai", 28)
]

columns = ["customer_id", "customer_name", "city", "age"]

df = spark.createDataFrame(data, columns)

df.show()
df.printSchema()
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("my_demo_catalog.demo_schema.customers")
```

```python
%sql
DESCRIBE detail my_demo_catalog.demo_schema.customers;
```

```python
%sql
DROP TABLE IF EXISTS demo_catalog.demo_schema.customers;
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .option("path", "abfss://mycontainer@demoblob1.dfs.core.windows.net/DemoSamples/customers/") \
    .saveAsTable("my_demo_catalog.demo_schema.customers")
```
