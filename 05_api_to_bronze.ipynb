```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

print(response.status_code)
```

```python
data = response.json()

print(data)
```

```python
import json

api_df = spark.createDataFrame(data)

display(api_df)
```

```python
api_df = spark.createDataFrame(data)
display(api_df)
```

```python
from pyspark.sql.functions import current_timestamp, lit

api_bronze = (
    api_df
    .withColumn(
        "_ingestion_timestamp",
        current_timestamp()
    )
    .withColumn(
        "_source",
        lit("jsonplaceholder_api")
    )
)
```

```python
(
    api_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "ingestion_autoloader.ingestion_demo.api_bronze"
    )
)
```
