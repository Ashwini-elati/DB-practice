```python
%sql
SELECT event_id, COUNT(*)
FROM streaming_demo.trades.stream_silver_trades
GROUP BY event_id
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    COUNT(*) AS total_events,
    COUNT(DISTINCT event_id) AS unique_events,
    MIN(event_time) AS earliest_event,
    MAX(event_time) AS latest_event,
    SUM(trade_value) AS total_trade_value
FROM streaming_demo.trades.stream_silver_trades;
```

```python
%sql
SELECT
    symbol,
    COUNT(*) AS trade_count,
    SUM(trade_value) AS total_value
FROM streaming_demo.trades.stream_silver_trades
GROUP BY symbol
ORDER BY total_value DESC;
```

```python
%sql
SELECT
    side,
    COUNT(*) AS trade_count,
    SUM(trade_value) AS total_value
FROM streaming_demo.trades.stream_silver_trades
GROUP BY side;
```

```python
%sql
SELECT
    window.start AS window_start,
    window.end AS window_end,
    symbol,
    trade_count,
    gross_value,
    buy_value,
    sell_value
FROM streaming_demo.trades.stream_gold_symbol_metrics
ORDER BY window_start, symbol;
```
