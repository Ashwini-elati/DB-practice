```python
%sql
-- =====================================================
-- 1. Bronze count
-- =====================================================

SELECT COUNT(*) AS bronze_count
FROM streaming_demo.trades.stream_bronze_trade_events;


-- =====================================================
-- 2. Silver count
-- =====================================================

SELECT COUNT(*) AS silver_count
FROM streaming_demo.trades.stream_silver_trades;


-- =====================================================
-- 3. Gold output
-- =====================================================

SELECT *
FROM streaming_demo.trades.stream_gold_symbol_metrics
ORDER BY window.start, symbol;


-- =====================================================
-- 4. Duplicate event IDs
-- =====================================================

SELECT
    event_id,
    COUNT(*) AS cnt
FROM streaming_demo.trades.stream_silver_trades
GROUP BY event_id
HAVING COUNT(*) > 1;


-- =====================================================
-- 5. Invalid side
-- =====================================================

SELECT COUNT(*) AS invalid_side_count
FROM streaming_demo.trades.stream_trade_quality_errors
WHERE quality_reason = 'INVALID_SIDE';


-- =====================================================
-- 6. Invalid quantity
-- =====================================================

SELECT COUNT(*) AS invalid_quantity_count
FROM streaming_demo.trades.stream_trade_quality_errors
WHERE quality_reason = 'INVALID_QUANTITY';


-- =====================================================
-- 7. Invalid price
-- =====================================================

SELECT COUNT(*) AS invalid_price_count
FROM streaming_demo.trades.stream_trade_quality_errors
WHERE quality_reason = 'INVALID_PRICE';


-- =====================================================
-- 8. Missing event ID
-- =====================================================

SELECT COUNT(*) AS missing_event_id_count
FROM streaming_demo.trades.stream_trade_quality_errors
WHERE quality_reason = 'MISSING_EVENT_ID';


-- =====================================================
-- 9. Bad timestamp
-- =====================================================

SELECT COUNT(*) AS invalid_timestamp_count
FROM streaming_demo.trades.stream_trade_quality_errors
WHERE quality_reason = 'INVALID_TIMESTAMP';


-- =====================================================
-- 10. Duplicate accounts
-- =====================================================

SELECT
    account_id,
    COUNT(*) AS cnt
FROM streaming_demo.trades.accounts
GROUP BY account_id
HAVING COUNT(*) > 1;
```
