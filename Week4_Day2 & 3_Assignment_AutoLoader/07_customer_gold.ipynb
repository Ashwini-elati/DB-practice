```python
customers = spark.table(
    "ingestion_autoloader.silver.customers"
)

api_customers = spark.table(
    "ingestion_autoloader.ingestion_demo.api_customer_silver"
)
```

```python
from pyspark.sql.functions import col

customer_gold = (
    customers.alias("c")
    .join(
        api_customers.alias("a"),
        col("c.customer_id") == col("a.customer_id"),
        "left"
    )
)

customer_gold = (
    api_customers
    .select(
        col("customer_id"),
        col("name").alias("customer_name"),
        col("email"),
        col("address.city").alias("city")
    )
)
```

```python
api_customers.printSchema()
```

```python
from pyspark.sql.functions import col

customer_gold = (
    api_customers
    .select(
        col("customer_id"),
        col("customer_name"),
        col("segment")
    )
)

display(customer_gold)
```

```python
(
    customer_gold
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "ingestion_autoloader.ingestion_demo.customer_gold"
    )
)
```
